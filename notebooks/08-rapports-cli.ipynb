# 08 — Lire les rapports déjà calculés

Pas de re-fit. On ouvre les JSON de `data/processed/` produits par le CLI
(`identify-r1c1`, `compare-r1c1-r2c2`, `mpc-cout-consigne`).

```python
import json
from basic_mpc.config import DataConfig

d = DataConfig().processed_dir
for name in (
    "r1c1_report.json",
    "r1c1_r2c2_report.json",
    "mpc_cout_consigne_report.json",
):
    path = d / name
    print("---", name, "existe" if path.is_file() else "absent")
    if not path.is_file():
        continue
    r = json.loads(path.read_text(encoding="utf-8"))
    if "params" in r and "tau_hours" in r["params"]:
        print("  τ R1C1 h", r["params"]["tau_hours"])
    if "params_r2c2" in r:
        p = r["params_r2c2"]
        print("  τ air / masse R2C2", p.get("tau_air_hours"), p.get("tau_mass_hours"))
    if "mpc" in r and "hysteresis" in r:
        print("  MPC h sous T_conf", r["mpc"].get("hours_under_conf"),
              "vs hyst", r["hysteresis"].get("hours_under_conf"))
```
