Code : `basic_mpc.models.impedance`, `basic_mpc.sim.payload`.

```python
from IPython.display import Image, display
from pathlib import Path
import tempfile

from basic_mpc.identification.plots_compare import plot_bode_phase, plot_z1_vectors
from basic_mpc.sim.payload import default_fitted_if_missing, vector_at_24h

r1, r2 = default_fitted_if_missing()
z = vector_at_24h(r1, r2)
print("R1C1 retard 24 h", round(z["r1c1"]["delay_hours"], 2), "h")
print("R2C2 retard 24 h", round(z["r2c2"]["delay_hours"], 2), "h")
tmp = Path(tempfile.mkdtemp())
plot_z1_vectors(r1, r2, tmp / "z1.png")
plot_bode_phase(r1, r2, tmp / "z3.png")
display(Image(str(tmp / "z1.png")))
display(Image(str(tmp / "z3.png")))
```
