## Constantes de temps

Sur le plant littérature : air ~3 h, masse ~11 h.
Sur le R2C2 **identifié** salon, les deux τ sont lentes (~140 h) — ce n'est pas ce plant.

```python
plant = literature_plant_params()
print("plant τ_air h =", plant.rae * plant.ca / 3600)
print("plant τ_mass h =", plant.ram * plant.cm / 3600)

r1 = R1C1Params(a=0.999, g_solar=0.0, g_heating=0.003)
print("R1C1 exemple τ_h =", r1.tau_hours)

r2 = R2C2Params(rae=5e5, ram=1.25e5, cm=4.27, g_solar=0.0, g_heating=1e-5)
print("R2C2 ident. τ_air / τ_mass h =", r2.tau_air_hours, r2.tau_mass_hours)
```
